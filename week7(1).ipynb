```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins = 20, kde=True)
plt.title("Histogram of Sepal Length")
plt.show()

sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.show()

species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts, labels=iris.target_names, autopct='%1.1f%%')
plt.title("Distribution of Species(pie chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue =df['species'])
plt.title("Scatterplot: Sepal length vs sepal width")
plt.show()

plt.figure(figsize = (6,4))
plt.plot(df['sepal length (cm)'])
plt.title("Line plot of Sepal Length")
plt.xlabel("Sample")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'],
            s = df['petal length (cm)']*20, alpha = 0.5,c=df['species'])
plt.title("Bubble Chart: Sepal vs Petal (size = petal length)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'],
            s = df['petal length (cm)']*20, alpha = 0.5, c = df['species'])
plt.title("Bubble Chart: Sepal vs petal (size = petal length)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title("Violin Plot: Sepal Length by Species")
plt.xticks([0,1,2], iris.target_names)
plt.show()
```

```python

tips = sns.load_dataset('tips')
df = tips
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=20, kde=True)

plt.title("Histogram of Total Bill")
plt.show()
sns.boxplot(x=df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.show()
species_counts = df['day'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.title("Distribution of Days (pie chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['day'])
plt.title("Scatterplot: Total Bill vs Tip")
plt.show()
plt.figure(figsize=(6,4))
plt.plot(df['total_bill'])
plt.title("Line plot of Total Bill")
plt.xlabel("Sample")
plt.ylabel("Total Bill")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['total_bill', 'tip', 'size']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'], df['tip'],
            s=df['tip']*20, alpha=0.5, c=df['size'])
plt.title("Bubble Chart: Total Bill vs Tip (size = tip)")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'], df['size'],
            s=df['tip']*20, alpha=0.5, c=df['tip'])
plt.title("Bubble Chart: Total Bill vs Size (size = tip)")
plt.xlabel("Total Bill")
plt.ylabel("Size")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(x=df['day'], y=df['total_bill'])
plt.title("Violin Plot: Total Bill by Day")
plt.show()
```

```python
from scipy import stats
b=np.array([120,122,118,130,125,128,115,121,123,119])
a=np.array([115,120,112,128,122,125,110,117,119,114])
alpha=0.05
t_stat,p_val=stats.ttest_rel(a,b)
m=np.mean(a-b)
s=np.std(a-b,ddof=1)
n=len(b)
t_manual=m/(s/np.sqrt(n))
decision="Reject" if p_val<=alpha else "Fail to reject"
concl="Significant difference." if decision == "Reject" else "No significant difference."
print("T:",t_stat)
print(f"Decision: {decision} H0 at a={alpha}")
print("Conclusion:",concl)
```

```python
from statsmodels.stats.weightstats import ztest
data = np.random.normal(loc=11.8, scale=0.5, size = 100)
population_mean = 12
population_std_dev=0.5
z_statistic,p_value=ztest(data, value=population_mean)
print(f"Z-statistic:{z_statistic}")
print(f"P-value:{p_value:.4f}")
alpha = 0.05
if p_value < alpha:
  print("Reject the null hypothesis: The average battery life is different from 12 hours.")
else:
  print("Fail to reject the null hypothesis: The average battery life is not significantly different from 12 hours")
```
